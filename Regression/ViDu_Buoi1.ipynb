```python

# Thư viện cần thiết

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
  # Tạo dữ liệu ngẫu nhiên
  # x là biến đầu vào, y là biến đầu ra
np.random.seed(0)
x = 2 * np.random.rand(100, 1)
y = 4 + 3 * x + np.random.randn(100, 1)
  # Đường thẳng với nhiễu
   # Chia dữ liệu thành tập huấn luyện và tập kiểm tra
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
  # Tạo mô hình hồi quy tuyến tính
model = LinearRegression()
  # Huấn luyện mô hình trên tập huấn luyện
model.fit(x_train, y_train)
  # Dự đoán trên tập kiểm tra
y_pred = model.predict(x_test)
  # Đánh giá mô hình
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
  # In kết quả
print(f"Hệ số chặn (Intercept): {model.intercept_}")
print(f"Hệ số góc (Slope): {model.coef_}")
print(f"Mean Squared Error (MSE): {mse}")
print(f"R^2 score: {r2}")
  # Vẽ biểu đồ
plt.scatter(x_test, y_test, color="blue", label="Dữ liệu thực")
plt.plot(x_test, y_pred, color="red", label="Dự đoán hồi quy")
plt.title("Hồi quy tuyến tính đơn giản")
plt.xlabel("Biến độc lập (x)")
plt.ylabel("Biến phụ thuộc (y)")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# --- PHẦN 1: Hồi quy tuyến tính đơn giản (Diện tích -> Giá bán) ---
np.random.seed(42)
n_samples_1 = 80

# Giả định diện tích từ 30m2 đến 150m2
area = 30 + 120 * np.random.rand(n_samples_1, 1)
# Công thức giả định: Giá bán = 15 + 1.8 * Diện tích + Nhiễu (triệu VND)
noise_1 = np.random.randn(n_samples_1, 1) * 20
price = 15 + 1.8 * area + noise_1

# Chia tập huấn luyện và kiểm tra
X_train_1, X_test_1, y_train_1, y_test_1 = train_test_split(area, price, test_size=0.2, random_state=42)

# Huấn luyện mô hình
model_simple = LinearRegression()
model_simple.fit(X_train_1, y_train_1)
y_pred_1 = model_simple.predict(X_test_1)

print("=== PHẦN 1: HỒI QUY TUYẾN TÍNH ĐƠN GIẢN ===")
print(f"Hệ số chặn (Intercept): {model_simple.intercept_[0]:.2f}")
print(f"Hệ số góc (Slope): {model_simple.coef_[0][0]:.2f}")
print(f"MSE: {mean_squared_error(y_test_1, y_pred_1):.2f}")
print(f"R2 Score: {r2_score(y_test_1, y_pred_1):.2f}\n")

# --- PHẦN 2: Thêm nhiều dữ liệu và tăng độ nhiễu ---
n_samples_2 = 1000
area_large = 30 + 120 * np.random.rand(n_samples_2, 1)
# Tăng độ nhiễu lên gấp đôi (độ lệch chuẩn = 40)
noise_large = np.random.randn(n_samples_2, 1) * 40
price_large = 15 + 1.8 * area_large + noise_large

# Chia dữ liệu
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(area_large, price_large, test_size=0.2, random_state=42)

# Huấn luyện mô hình
model_noisy = LinearRegression()
model_noisy.fit(X_train_2, y_train_2)
y_pred_2 = model_noisy.predict(X_test_2)

print("=== PHẦN 2: DỮ LIỆU LỚN HƠN & ĐỘ NHIỄU CAO ===")
print(f"Hệ số chặn (Intercept): {model_noisy.intercept_[0]:.2f}")
print(f"Hệ số góc (Slope): {model_noisy.coef_[0][0]:.2f}")
print(f"MSE: {mean_squared_error(y_test_2, y_pred_2):.2f}")
print(f"R2 Score: {r2_score(y_test_2, y_pred_2):.2f}\n")

# --- PHẦN 3: Hồi quy tuyến tính đa biến (Thêm đặc trưng số phòng và khoảng cách) ---
# Sinh thêm các biến độc lập mới
# 1. Số phòng ngủ: từ 1 đến 5 phòng
rooms = np.random.randint(1, 6, size=(n_samples_2, 1))
# 2. Khoảng cách tới trung tâm (km): từ 1km đến 20km
distance = 1 + 19 * np.random.rand(n_samples_2, 1)

# Công thức giả định giá bán phức tạp hơn:
# Giá = 15 + 1.5 * Diện tích + 12 * Số phòng - 3 * Khoảng cách + Nhiễu
price_multi = 15 + 1.5 * area_large + 12 * rooms - 3 * distance + np.random.randn(n_samples_2, 1) * 15

# Gộp các đặc trưng thành một ma trận X
X_multi = np.hstack((area_large, rooms, distance))

# Chia dữ liệu
X_train_3, X_test_3, y_train_3, y_test_3 = train_test_split(X_multi, price_multi, test_size=0.2, random_state=42)

# Huấn luyện mô hình đa biến
model_multi = LinearRegression()
model_multi.fit(X_train_3, y_train_3)
y_pred_3 = model_multi.predict(X_test_3)

print("=== PHẦN 3: HỒI QUY TUYẾN TÍNH ĐA BIẾN ===")
print(f"Hệ số chặn (Intercept): {model_multi.intercept_[0]:.2f}")
print(f"Hệ số góc (Coefficients) [Diện tích, Số phòng, Khoảng cách]: {model_multi.coef_[0]}")
print(f"MSE: {mean_squared_error(y_test_3, y_pred_3):.2f}")
print(f"R2 Score: {r2_score(y_test_3, y_pred_3):.2f}\n")

# --- Trực quan hóa kết quả so sánh giữa mô hình đơn giản và dữ liệu nhiễu ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Biểu đồ mô hình đơn giản
ax1.scatter(X_test_1, y_test_1, color='blue', alpha=0.7, label='Dữ liệu thực tế')
ax1.plot(X_test_1, y_pred_1, color='red', linewidth=2, label='Đường hồi quy')
ax1.set_title('Hồi quy Đơn giản (Ít mẫu, Nhiễu thấp)')
ax1.set_xlabel('Diện tích (m2)')
ax1.set_ylabel('Giá bán (triệu VND)')
ax1.legend()
ax1.grid(True)

# Biểu đồ mô hình nhiều nhiễu
ax2.scatter(X_test_2, y_test_2, color='green', alpha=0.4, label='Dữ liệu thực tế')
ax2.plot(X_test_2, y_pred_2, color='red', linewidth=2, label='Đường hồi quy')
ax2.set_title('Hồi quy Đơn giản (Nhiều mẫu, Nhiễu cao)')
ax2.set_xlabel('Diện tích (m2)')
ax2.set_ylabel('Giá bán (triệu VND)')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()
```

### 📋 Hướng dẫn các bước xây dựng mô hình Hồi quy tuyến tính từ cơ bản đến nâng cao

Để giải quyết bài toán dự đoán giá nhà bằng hồi quy tuyến tính, chúng ta nên thực hiện tuần tự theo 5 bước tiêu chuẩn sau:

#### **Bước 1: Chuẩn bị Thư viện**
Chúng ta cần import các công cụ cần thiết để làm việc với dữ liệu, xây dựng mô hình và vẽ đồ thị:
* `numpy`: Để sinh và tính toán dữ liệu dạng ma trận/mảng.
* `matplotlib.pyplot`: Để vẽ biểu đồ trực quan hóa dữ liệu và đường hồi quy.
* `sklearn.linear_model.LinearRegression`: Thư viện chứa mô hình hồi quy tuyến tính.
* `sklearn.model_selection.train_test_split`: Hàm giúp chia bộ dữ liệu thành tập huấn luyện (Train) và tập kiểm tra (Test).
* `sklearn.metrics`: Các hàm đánh giá sai số như `mean_squared_error` (MSE) và độ khớp `r2_score`.

#### **Bước 2: Chuẩn bị Dữ liệu (Sinh dữ liệu giả định)**
Do chưa có file dữ liệu thực tế, ta tự sinh dữ liệu mô phỏng:
* **Khởi tạo Seed (`np.random.seed(42)`)**: Giúp kết quả ngẫu nhiên luôn cố định ở mọi lần chạy.
* **Xây dựng biến độc lập (X)**: Diện tích nhà (ví dụ chạy ngẫu nhiên từ 30m² đến 150m²).
* **Xây dựng biến phụ thuộc (y)**: Tính Giá bán dựa theo công thức thực tế giả định cộng thêm một lượng sai số ngẫu nhiên (nhiễu/noise) để mô phỏng thực tế.

#### **Bước 3: Chia tập dữ liệu (Train - Test Split)**
* Chia ngẫu nhiên dữ liệu thành 2 phần: **Tập huấn luyện (Train)** dùng để máy học (thường chiếm 80%) và **Tập kiểm tra (Test)** dùng để đánh giá độ chính xác (chiếm 20%).
* Việc này tránh hiện tượng mô hình chỉ học vẹt trên dữ liệu cũ mà không dự đoán được dữ liệu mới.

#### **Bước 4: Huấn luyện Mô hình (Fit Model)**
* Khởi tạo đối tượng mô hình: `model = LinearRegression()`
* Truyền dữ liệu huấn luyện vào để mô hình tìm ra hệ số chặn (Intercept) và hệ số góc (Coefficients) tối ưu nhất: `model.fit(X_train, y_train)`

#### **Bước 5: Dự đoán và Đánh giá (Predict & Evaluate)**
* Dùng mô hình đã học để dự đoán giá bán trên tập kiểm tra: `y_pred = model.predict(X_test)`
* Đo lường hiệu năng bằng cách so sánh giá trị dự đoán `y_pred` với giá trị thực tế `y_test` thông qua chỉ số **MSE** (càng thấp càng tốt) và **R² Score** (càng gần 1 càng tốt).
* Vẽ biểu đồ Scatter (điểm dữ liệu thực tế) kết hợp với Line (đường thẳng dự đoán) để trực quan hóa kết quả.
